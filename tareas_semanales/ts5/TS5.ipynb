## Circuito 1 (Red T)

![redT](images/redT.png)

La obtencion de la MAI con solo 3 bornes se llega a complicar bastante debido a que se tiene quer realizar admitancias en paralelo y serie. Debido a eso se opto por agregar un borne mas entre 1 y 2 (borne 4), con este borne extra es mucho mas facil obtener la matriz MAI 4x4, luego se realiza una reduccion de el borne 4, de esta manera volviendo al ejercicio original.

![redT2](images/redT2.png)

Para armar la MAI se la realiza como si fuera una matriz admitancia.

$$ Y_{MAI} = \begin{pmatrix} \frac{1}{SL} & 0 & 0 & -\frac{1}{SL} \\
                            0 & \frac{1}{SL} & 0 & -\frac{1}{SL} \\
                            0 & 0 & SC & -SC \\
                            -\frac{1}{SL} & -\frac{1}{SL} & -SC & \frac{2}{SL}+SC \end{pmatrix}
                            $$

Para obtener la MAI de 3x3:

$$Y_{MAI}^{3x3} = Y_{11} - Y_{12} \cdot Y_{22}^{-1} \cdot Y_{21}$$

$$Y_{11} = \begin{pmatrix} \frac{1}{SL} & 0 & 0 \\ 0 & \frac{1}{SL} & 0 \\ 0 & 0 & SC \end{pmatrix} \hspace{2cm} Y_{12} = \begin{pmatrix} -\frac{1}{SL} \\ -\frac{1}{SL} \\ -SC \end{pmatrix} \hspace{2cm}  Y_{22} = \frac{2}{SL} + SC \hspace{2cm} Y_{21} = \begin{pmatrix} -\frac{1}{SL} & -\frac{1}{SL} & -SC \end{pmatrix}$$

$$Y_{MAI}^{3x3} =  \begin{pmatrix} \frac{1}{SL} & 0 & 0 \\ 0 & \frac{1}{SL} & 0 \\ 0 & 0 & SC\end{pmatrix} - \begin{pmatrix} -\frac{1}{SL} \\ -\frac{1}{SL} \\ -SC \end{pmatrix} \cdot \frac{1}{2/SL+SC} \cdot \begin{pmatrix} -\frac{1}{SL} & -\frac{1}{SL} & -SC \end{pmatrix}$$

Operando se puede llegar al siguiente resultado.

$$Y_{MAI}^{3x3} = \frac{1}{2/SL + SC} \cdot \begin{pmatrix} \frac{1}{SL}(\frac{1}{SL} + SC) & -\frac{1}{S^2L^2} & -\frac{C}{L} \\ -\frac{1}{S^2L^2} & \frac{1}{SL}(\frac{1}{SL} + SC) & -\frac{C}{L} \\ -\frac{C}{L} & -\frac{C}{L} & \frac{2C}{L}\end{pmatrix}$$

se puede observar que cumple las propiedades de MAI, como que la suma de filas o columnas es igual a 0.


Obtener la ganancia para $\frac{V_{23}}{V_{13}} \hspace{1cm} \frac{V_{21}}{V_{31}} \hspace{1cm} \frac{V_{12}}{V_{32}}$

Se puede obtenerlo de la siguiente manera:

$$\frac{V_{23}}{V_{13}} = (-1)^{2 + 3 + 1 + 3} \cdot \frac{Y_{23}^{13}}{Y^1_{3}}$$

En el cofactor $Y^1_{3}$ se elimina fila 1 y columna 3 de la matriz MAI.

$$Y^1_{3}= \begin{pmatrix} -\frac{1}{S^2L^2} & \frac{1}{SL}(\frac{1}{SL} + SC) \\ -\frac{C}{L} & -\frac{C}{L} \end{pmatrix}$$

sacando el determinante.

$$\Delta Y^1_{3} = \frac{C}{SL^2}(\frac{2}{SL} + SC)$$

En el cofactor $Y_{23}^{13}$ se eliman filas 2 y 3, y columnas 1 y 3.

$$Y_{23}^{13} = -\frac{1}{S^2L^2}$$

$$ \boxed{\frac{V_{23}}{V_{13}} = -1 \cdot \frac{-\frac{1}{S^2L^2}}{\frac{C}{SL^2}(\frac{2}{SL} + SC)} = \frac{\frac{1}{C^2}}{S^2 + \frac{2}{LC}}}$$



Operando de la misma manera para los otros casos.

$$\frac{V_{21}}{V_{31}} = (-1)^{2 + 1 + 3 + 1} \cdot \frac{Y_{21}^{31}}{Y^3_{1}}$$

Quedando el cofactor $Y^3_{1}$

$$Y^3_{1} = \begin{pmatrix} -\frac{1}{SL^2} & -\frac{C}{L} \\ -\frac{1}{\frac{1}{SL}(\frac{1}{SL}+SC)} & -\frac{C}{L} \end{pmatrix}$$

$$\Delta Y^3_{1} = \frac{C}{SL^2}(\frac{2}{SL} + SC)$$

El cofactor $Y_{23}^{13} = -\frac{C}{L}$ 

$$ \boxed{\frac{V_{12}}{V_{32}} = -1 \cdot \frac{-\frac{1}{S^2L^2}}{\frac{C}{SL^2}(\frac{2}{SL} + SC)} = \frac{S^2\frac{L}{C}}{S^2 + \frac{2}{LC}}}$$



Y para el ultimo caso.

$$\frac{V_{21}}{V_{31}} = (-1)^{1 + 2 + 3 + 2} \cdot \frac{Y_{12}^{32}}{Y^3_{2}}$$

Quedando el cofactor $Y^3_{2}$

$$Y^3_{2} = \begin{pmatrix} \frac{1}{SL}(\frac{1}{SL} + SC) & -\frac{C}{L} \\ -\frac{1}{S^2L^2} & -\frac{C}{L} \end{pmatrix}$$

$$\Delta Y^3_{2} = -\frac{C}{SL^2}(\frac{2}{SL} + SC)$$

El cofactor $Y_{32}^{12} = -\frac{C}{L}$ 

$$ \boxed{\frac{V_{12}}{V_{32}} = \frac{-\frac{1}{S^2L^2}}{-\frac{C}{SL^2}(\frac{2}{SL} + SC)} = \frac{S^2\frac{L}{C}}{S^2 + \frac{2}{LC}}}$$


Teniendo los siguientes resultados.

$$\boxed{T_1 = \frac{\frac{1}{C^2}}{S^2 + \frac{2}{LC}} \hspace{1cm} T_2 = \frac{S^2\frac{L}{C}}{S^2 + \frac{2}{LC}} \hspace{1cm} T_3 = \frac{S^2\frac{L}{C}}{S^2 + \frac{2}{LC}}}$$


```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as sig
L = 1
C = 1

num1 = [1/(C**2)]
den1 = [1, 0, 2/(L*C)]

num2 = [L/C, 0, 0]
den2 = [1, 0, 2/(L*C)]

num3 = [L/C, 0, 0]
den3 = [1, 0, 2/(L*C)]

#armado de la transferencia
worn = np.logspace(-1,2,1000)
w1, h1 = sig.freqs(num1, den1, worN=worn)
w2, h2 = sig.freqs(num2, den2, worN=worn)
w3, h3 = sig.freqs(num3, den3, worN=worn)


plt.figure(figsize=(12,6))

plt.subplot(2,1,1)
plt.semilogx(w1, 20*np.log10(np.abs(h1)))
plt.semilogx(w2, 20*np.log10(np.abs(h2)))
plt.semilogx(w3, 20*np.log10(np.abs(h3)), "--")
plt.ylabel("Amplitud [dB]")
plt.xlabel("frecuencia [rad/s]")
plt.legend(['T1', 'T2', 'T3'])
plt.tight_layout()
plt.grid()

plt.subplot(2,1,2)
plt.semilogx(w1, np.angle(h1)*180/np.pi)
plt.semilogx(w2, np.angle(h2)*180/np.pi)
plt.semilogx(w3, np.angle(h3)*180/np.pi, "--")
plt.ylabel("Fase [grados]")
plt.xlabel("frecuencia [rad/s]")
plt.legend(['T1', 'T2', 'T3'])
plt.tight_layout()
plt.grid()
```

```python

numt = [L/C, 0, 1/C**2]
dent = [1, 0, 2/(L*C)]

w, h = sig.freqs(numt, dent, worN=worn)

plt.figure(figsize=(12,6))
plt.subplot(2,1,1)
plt.semilogx(w, 20*np.log10(np.abs(h)))
plt.ylabel("Amplitud [dB]")
plt.xlabel("frecuencia [rad/s]")
plt.grid()

plt.subplot(2,1,2)
plt.semilogx(w, np.angle(h)*180/np.pi)
plt.ylabel("Fase [grados]")
plt.xlabel("frecuencia [rad/s]")
plt.grid()

plt.tight_layout()
plt.show()
```

Para el tema de los graficos se realizara en python y se tomara como referencia el normalizado.

Se puede observar en los graficos que en primera cuestion parecen complementarios T1 con T2 y T3, pero no del todo debido a que T1 para frecuencias bajas atenua y no mantiene, debido a esto se obtiene que no son complementarias en sentido estricto, en bajas frecuencias domina T1 y en altas frecuencias domina T2 no son complementarios plenamente T1+T2 es distinto de 1.

Tiene la misma frecuencia de corte y la fase es casi la misma en el caso de T1 la fase solo es aportada por los polos, en cambio en T2 y T3 se aporta tambien la fase de los ceros como se observa en los graficos.

## Circuito 2 (MOS)

![MOS](images/MOS.png)

Realizamos un circuito equivalente al cual deberiamos de llegar.

![MOS1](images/MOS_nogene.png)

$$C_{iss}= C_{gs} + C{gd} \qquad C_{oss} = C_{ds} + C_{gd} \qquad C_{rss}=C_{gd}$$


Para poder realizar la matriz MAI del MOS podemos proceder con el circuito equivalente del MOS.

![MOS2](images/MOS_gene.png)

Para el armado de la matriz MAI podemos realizarlo con superposicion, de manera que podemos corto circuitar el generador que se tiene, de esta manera podemos obtener la "Y" de los pasivos.

$$Y_{pasivos} = \begin{pmatrix} SC_{gd} + SC{gs} & -SC{gd} & -SC{ds} \\ -SC_{gd} & SC_{gd}+SC_{ds} & -SC_{ds} \\ -SC_{gs} & -SC{ds} & SC{gs} + SC{ds} \end{pmatrix} = \begin{pmatrix} S(C_{gd} + C{gs}) & -SC{gd} & -SC{ds} \\ -SC_{gd} & S(C_{gd}+C_{ds}) & -SC_{ds} \\ -SC_{gs} & -SC{ds} & S(C{gs} + C{ds}) \end{pmatrix}$$

Luego procedemos con el generador controlado.

Sabiendo que la corriente de la fuente va de 2 a 3 la corriente que entra en el nodo 2 es $i_d$.

$$i_d=g_mV_gs \quad con \quad V_gs=V_1-V_3 \Rightarrow \boxed{i_2=i_d=g_m(V_1-V_3)}$$

En el nodo 3 nos queda lo contrario:

$$i_3=-g_mV_gs \quad con \quad V_gs=V_1-V_3 \Rightarrow \boxed{i_3=i_d=-g_m(V_1-V_3)}$$

$$i = Y\cdot v$$

$$\boxed{Y_{gm} = \begin{pmatrix} 0 & 0 & 0 \\ g_m & 0 & -g_m \\ -g_m & 0 & g_m \end{pmatrix}}$$

Sumando ambas Y debido a superposicion nos queda de la siguiente manera:

$$Y_{MAI} = Y_{pasicos} + Y_{gm} = \begin{pmatrix} S(C_{gd} + C{gs}) & -SC{gd} & -SC{ds} \\ g_m-SC_{gd} & S(C_{gd}+C_{ds}) & -g_m-SC_{ds} \\ -g_m-SC_{gs} & -SC{ds} & g_m +S(C{gs} + C{ds}) \end{pmatrix}$$

Ahora volviendo al caso de los parametros que se debe de llegar reemplazamos en la matriz obtenida:


$$C_{iss}= C_{gs} + C{gd} \qquad C_{oss} = C_{ds} + C_{gd} \qquad C_{rss}=C_{gd}$$

$$Y_{MAI} = \begin{pmatrix} SC_{iss} & -SC{rss} & -S(C{iss}-C{rss}) \\ g_m-SC_{rss} & SC_{oss} & -g_m-S(C_{oss}-C_{rss}) \\ -g_m-S(C_{iss}-C{rss}) & -SC{oss} & g_m +S(C{iss} + C{oss} - 2C{rss}) \end{pmatrix}$$

La matriz admitancia indefenida del MOS.

Para obtener la matriz Y de un drain comun, como se tiene la matriz $Y_{MAI}$ podemos aterrizar la matriz MAI y asi obtener la matriz Y de un drain comun o cualquier configuracion, eliminando la fila y columna correspondiente a la configuracion que se requiera.

DRAIN COMUN.

Aterrizamos la matriz MAI eliminando la fila 2 y columna 2.

$$Y_{DC} = \begin{pmatrix} SC_{iss} & -S(C_i{ss}-C_{rss})\\-g_m-S(C_{iss}-C_{rss}) & g_m +S(C{iss} + C{oss} - 2C{rss})  \end{pmatrix}$$

SOURCE COMUN

Aterrizamoas la matriz MAI, eliminando la fila 3 y columna 3.

$$Y_{SC} = \begin{pmatrix} SC_{iss} & -SC_{rss} \\ g_m-SC_{rss} & SC{oss} \end{pmatrix}$$

Para el caso de querer obtener los parametros ABCD en la configuracion source comun, como tenemos la matriz Y, se puede realizar directamente transforma de Y a ABCD.

$$Y \Rightarrow T \qquad \begin{pmatrix} Y_1 & Y_2 \\ Y_3 & Y_4 \end{pmatrix} \Rightarrow \begin{pmatrix} -Y_4/Y_3 & -1/Y3 \\ \Delta Y/Y3 & -Y1/Y3 \end{pmatrix}$$

$$\Delta Y = (SC_{iss}\cdot SC{oss})-(-SC_{rss})(g_m-SC_{rss}) = S^2(C_{iss}\cdot C_{oss} - C_{rss}^2) + S C_{rss}g_m$$

$$\Delta Y = S[S(C_{iss}\cdot C_{oss} - C_{rss}^2) + C_{rss}g_m]$$

$$\boxed{\begin{pmatrix} -\frac{SC{oss}}{g_m-SC_{rss}} & -\frac{1}{g_m-SC_{rss}} \\  \frac{S[S(C_{iss}\cdot C_{oss} - C_{rss}^2) + C_{rss}g_m]}{g_m-SC_{rss}} & -\frac{SC_{iss}}{g_m-SC_{rss}}\end{pmatrix}}$$
